```python
import json

json_data = {}
json_data['designs'] = []
for i, device in enumerate(devices):
    temp = {
        "name": f"design-{i+1}",
        "period": device.ax,
        "period_taper": device.d_max,
        "hole_rad_x": device.rx,
        "hole_rad_y": device.ry,
        "hole_rad_x_taper": device.d_max_rx,
        "hole_rad_y_taper": device.d_max_ry,
        "cavity_hole_num": device.cavity_hole_num
    }
    json_data['designs'].append(temp)
with open("tantala_pcc_designs.json", "w") as f:
    json.dump(json_data, f, indent=4)
```

```python
import gdstk

# 1. Load the existing GDS file
lib_loaded = gdstk.read_gds("./coupler_old.gds")

for cell in lib_loaded.cells:
    print(cell.name)
# 2. Get the specific cell and extract its polygons
source_cell = lib_loaded.cells[0]  # or use a dict/name lookup
polygons = source_cell.get_polygons()

# 3. Create a new library and cell for your output
new_lib = gdstk.Library()
target_cell = new_lib.new_cell("MAIN")

# 4. Copy existing shapes, assign a new layer (e.g., layer=5, datatype=0)
for poly in polygons:
  print(poly)
  new_poly = poly.copy()
  new_poly.layer = 1
  new_poly.datatype = 0
  target_cell.add(new_poly)

# 5. Save the new GDS file
new_lib.write_gds("test.gds")
```
