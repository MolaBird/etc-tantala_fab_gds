```python
import gdstk
import math

# 1. Start with the straight edges: (1, 0) to (0, 0)
points = [(1, 0), (0, 0)]

# 2. Generate the inward-bending arc from (0, 1) to (1, 0)
# The center of this arc is at (1, 1) with a radius of 1.
# The angle sweeps from pi (180 degrees) to 1.5 * pi (270 degrees).
num_arc_points = 64
for i in range(num_arc_points + 1):
    angle = math.pi + (math.pi / 2) * (i / num_arc_points)
    x = 1 + math.cos(angle)
    y = 1 + math.sin(angle)
    points.append((x, y))

# 3. Create the polygon from the points
enclosure = gdstk.Polygon(points)

# 4. Create a cell and add the polygon to it
cell = gdstk.Cell("INWARD_ENCLOSURE")
cell.add(enclosure)

# 5. Create a library, add the cell, and save the GDS file
lib = gdstk.Library()
lib.add(cell)
lib.write_gds("test.gds")
```
